```python
import scipy.cluster.hierarchy as sch
val = [[2,5],[3,6],[7,10],[12,10],[6,3],[10,20]]
sch.linkage(val)
```

```python
import matplotlib.pyplot as plt
dendrogram = sch.dendrogram(sch.linkage(val,method='single'))
plt.title("Dendrogram")
plt.xlabel("Clusters")
plt.ylabel("Height")
plt.show()
```

```python
import pandas as pd
import numpy as np
val = [[2,5],[3,6],[7,10],[12,10],[6,3],[10,20]]
data=[]
for i in range(len(val)):
    for j in range(len(val)):
        p1=val[i]
        p2=val[j]
        
        # priting Distance between point p1 and p2, (p1&p2 will change for every 'for'loop iteration)
        
        dis = np.sqrt((p2[0]-p1[0])**2 + (p2[1]-p1[1])**2)
        data.append([p1,p2,dis])
pd.DataFrame(data)
        
```
