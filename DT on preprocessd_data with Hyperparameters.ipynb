```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

import warnings
warnings.simplefilter("ignore")
```

```python
X = df.drop("churn",axis=1)
print(X)
print("="*100)

y = df["churn"]
print(y)
print("="*100)

from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,train_size=0.8,random_state=6)

from sklearn.tree import DecisionTreeClassifier
model = DecisionTreeClassifier()
model.fit(X_train,y_train)
```

```python
ypred_train = model.predict(X_train)
print(ypred_train)
ypred_test = model.predict(X_test)
print(ypred_test)

```

```python
train_accuracy = model.score(X_train,y_train)
print("train_accuracy",train_accuracy)
test_accuracy = model.score(X_test,y_test)
print("test_accuracy",test_accuracy)

from sklearn.model_selection import cross_val_score
scores = cross_val_score(model,X,y,cv=5)
print("cross_val_score",scores.mean())
```

```python
from sklearn.tree import plot_tree 

plt.figure(figsize=(20,24),dpi=300)   # It will increses the Decision Tree size.

plot_tree(model,filled=True,feature_names=list(X_train.columns),
         #eature_names=["gender", "age", "no_of_days_subscribed", "multi_screen mail_subscribed","weekly_mins_watched","minimum_daily_mins","weekly_max_night_mins" ,"videos_watched" ,"customer_support_calls"],
          class_names=["0","1"])
plt.show()
          
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.tree import DecisionTreeClassifier

estimator = DecisionTreeClassifier()
grid_param = {"criterion":["gini","entropy"],"max_depth":[2,3,4,5,6],"max_leaf_nodes":[1,2,3]}

dt_classifier = GridSearchCV(estimator,grid_param,cv=5,scoring="accuracy",verbose=True)
dt_classifier.fit(X_train,y_train)

dt_classifier.best_params_
```

```python
dt_classifier.best_estimator_
```

```python
dt_classifier.best_score_
```

```python
import time
start = time.time()
dt_classifier.fit(X_train,y_train)
end = time.time()
print("Total time taken to fit the model is:",(end-start))
```
