```python
df["genres"] = df["genres"].fillna('')   #--> replacing the null values with null string
```

```python
# TF-IDF Vectorization :  It will covert the text words into the numerical formate by creating vector(matrix) for senteces.
                        # It will give the importance value of each word from sentence in matrix formate(means it will give check how many times a word is there in sentence for each unique word and depending on that it will consider the importnace of that word.)

from sklearn.feature_extraction.text import TfidfVectorizer
tfidfv = TfidfVectorizer()
feature_vectors = tfidfv.fit_transform(df["genres"])
print(feature_vectors)
```

#### Cosine Similarity :

```python
from sklearn.metrics.pairwise import cosine_similarity
similarity = cosine_similarity(feature_vectors)
print(similarity)
```

```python
similarity.shape
```

```python
list_of_all_titles = df['title'].tolist()
print(list_of_all_titles)
```

```python
# getting the movie name from the user
movie_name = input("Enter your favourite movie name : ")
```

```python
# finding the close match for the movie name given by the user

import difflib
find_close_match = difflib.get_close_matches(movie_name, list_of_all_titles)
print(find_close_match)
```

```python
close_match = find_close_match[0]
print(close_match)
```

```python
# finding the index of the movie with title

index_of_the_movie = df[df.title == close_match]['index'].values[0]
print(index_of_the_movie)
```

```python
# getting a list of similar movies

similarity_score = list(enumerate(similarity[index_of_the_movie]))
similarity_score
```

```python
len(similarity_score)
```

```python
# sorting the movies based on their similarity score

sorted_similar_movies = sorted(similarity_score, key = lambda x:x[1], reverse = True)
sorted_similar_movies
```

#### Printing the name of similar movies based on the index

```python
print('Movies suggested for you : \n')
i = 1
for movie in sorted_similar_movies : 
    index = movie[0]
    title_from_index = df[df.index==index]['title'].values[0]
    if (i<=10):
        print(i,'.',title_from_index)
    i+=1
```
